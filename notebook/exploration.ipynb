## Carregamento dos dados metabolômicos

## Seleção das variáveis clínicas relevantes

Nem todas as variáveis presentes na base original são necessárias para a construção do conjunto de dados final. Dessa forma, realiza-se uma etapa de filtragem, mantendo apenas os atributos considerados relevantes para a análise, incluindo características demográficas, informações tumorais e fatores clínicos potencialmente relacionados ao problema investigado.

```python
cols_importantes = [
    'Source Name',
    'Factor Value[Sex]',
    'Factor Value[Tumor state]',
    'Factor Value[Tumor grade]',
    'Factor Value[Age]',
    'Factor Value[Neutral Status]',
    'Factor Value[ER status]',
    'Factor Value[HER2 status]',
    'Factor Value[Intercurrent tumors]',
    'Factor Value[Breed]'
]
```

```python
samples_clean = samples[cols_importantes]
```

## Padronização dos nomes das variáveis

Após a seleção dos atributos relevantes, os nomes das colunas são padronizados com o objetivo de facilitar a manipulação computacional e melhorar a legibilidade do código. Essa etapa reduz inconsistências relacionadas a espaços, caracteres especiais e nomenclaturas extensas, tornando o processo de análise mais organizado.


```python
samples_clean = samples_clean.rename(columns={
    'Source Name': 'sample_id',
    'Factor Value[Sex]': 'sex',
    'Factor Value[Tumor state]': 'tumor_state',
    'Factor Value[Tumor grade]': 'tumor_grade',
    'Factor Value[Age]': 'age',
    'Factor Value[Neutral Status]': 'neutered',
    'Factor Value[ER status]': 'er_status',
    'Factor Value[HER2 status]': 'her2_status',
    'Factor Value[Intercurrent tumors]': 'tumor_type',
    'Factor Value[Breed]': 'breed'
})
```

## Transformação da base metabolômica

A estrutura original da base metabolômica encontra-se organizada de maneira distinta da necessária para os algoritmos de aprendizado de máquina. Assim, é realizada uma transformação dos dados, reorganizando os metabólitos e amostras em um formato tabular adequado, no qual cada linha representa uma amostra e cada coluna representa um metabólito identificado.

## Seleção das colunas metabolômicas relevantes

A base metabolômica original contém informações descritivas sobre os metabólitos, como fórmula química, identificadores e outros metadados. Entretanto, para a construção do conjunto de dados utilizado em aprendizado de máquina, são necessárias apenas as variáveis correspondentes aos valores metabolômicos das amostras e o identificador de cada metabólito.

Assim, nesta etapa são selecionadas apenas a coluna de identificação dos metabólitos e as colunas referentes às amostras experimentais, identificadas pelo padrão `_CrAdj`.

```python
# Identifica automaticamente as colunas das amostras
sample_cols = [c for c in metabolites.columns if "_CrAdj" in c]

# Mantém apenas o identificador do metabólito e as colunas das amostras
metabolites_ml = metabolites[
    ['metabolite_identification'] + sample_cols
]

metabolites_ml.head()
```

## Definição do identificador dos metabólitos

Após a seleção das variáveis relevantes, a coluna responsável pela identificação dos metabólitos é definida como índice da tabela. Essa etapa é necessária porque, posteriormente, será realizada uma reorganização estrutural dos dados, transformando metabólitos em colunas e amostras em linhas.

```python
# Define o nome do metabólito como índice da tabela
metabolites_ml = metabolites_ml.set_index(
    'metabolite_identification'
)

metabolites_ml.head()
```

## Reorganização estrutural da base metabolômica

A base metabolômica encontra-se originalmente estruturada com metabólitos organizados em linhas e amostras distribuídas em colunas. No entanto, algoritmos de aprendizado de máquina exigem uma estrutura em que cada linha represente uma amostra e cada coluna represente uma variável preditora.

Dessa forma, realiza-se a transposição da tabela, invertendo linhas e colunas.

```python
# Inverte linhas e colunas da tabela
metabolites_ml = metabolites_ml.T

metabolites_ml.head()
```

## Conversão do identificador das amostras em coluna

Após a transposição da tabela, os identificadores das amostras passam a ocupar o índice da estrutura de dados. Entretanto, para permitir a integração com a base clínica, esse índice deve ser convertido em uma coluna explícita, denominada `sample_id`.

```python
# Converte o índice em coluna
metabolites_ml = metabolites_ml.reset_index()

# Renomeia a coluna de índice
metabolites_ml = metabolites_ml.rename(
    columns={'index': 'sample_id'}
)

metabolites_ml.head()
```

## Padronização dos identificadores das amostras

Os identificadores presentes na base metabolômica possuem o sufixo `_CrAdj`, enquanto a base clínica utiliza apenas o identificador principal da amostra. Assim, remove-se esse trecho para garantir compatibilidade entre os conjuntos de dados.

```python
# Remove o sufixo '_CrAdj' dos identificadores
metabolites_ml['sample_id'] = (
    metabolites_ml['sample_id']
    .str.replace('_CrAdj', '', regex=False)
)

metabolites_ml.head()
```

## Integração das bases clínica e metabolômica

Após o pré-processamento das informações clínicas e metabolômicas, realiza-se a integração das bases utilizando o identificador único da amostra (`sample_id`). Esse procedimento permite consolidar, em uma única estrutura, tanto os atributos clínicos quanto os valores metabolômicos.

É utilizada uma junção interna (`inner join`), garantindo que apenas amostras presentes simultaneamente nas duas bases sejam mantidas no conjunto final.

```python
# Integra as bases clínica e metabolômica
dataset_final = pd.merge(
    samples_clean,
    metabolites_ml,
    on="sample_id",
    how="inner"
)

print(dataset_final.shape)
dataset_final.head()
```
