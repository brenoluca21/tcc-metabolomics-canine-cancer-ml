## Transformação da base metabolômica

A estrutura original da base metabolômica encontra-se organizada de maneira distinta da necessária para os algoritmos de aprendizado de máquina. Assim, é realizada uma transformação dos dados, reorganizando os metabólitos e amostras em um formato tabular adequado, no qual cada linha representa uma amostra e cada coluna representa um metabólito identificado.

## Seleção das colunas metabolômicas relevantes

A base metabolômica original contém informações descritivas sobre os metabólitos, como fórmula química, identificadores e outros metadados. Entretanto, para a construção do conjunto de dados utilizado em aprendizado de máquina, são necessárias apenas as variáveis correspondentes aos valores metabolômicos das amostras e o identificador de cada metabólito.

Assim, nesta etapa são selecionadas apenas a coluna de identificação dos metabólitos e as colunas referentes às amostras experimentais, identificadas pelo padrão `_CrAdj`.

```python
# Identifica automaticamente as colunas das amostras
sample_cols = [c for c in metabolites.columns if "_CrAdj" in c]

# Mantém apenas o identificador do metabólito e as colunas das amostras
metabolites_ml = metabolites[
    ['metabolite_identification'] + sample_cols
]

metabolites_ml.head()
```

## Definição do identificador dos metabólitos

Após a seleção das variáveis relevantes, a coluna responsável pela identificação dos metabólitos é definida como índice da tabela. Essa etapa é necessária porque, posteriormente, será realizada uma reorganização estrutural dos dados, transformando metabólitos em colunas e amostras em linhas.

```python
# Define o nome do metabólito como índice da tabela
metabolites_ml = metabolites_ml.set_index(
    'metabolite_identification'
)

metabolites_ml.head()
```

## Reorganização estrutural da base metabolômica

A base metabolômica encontra-se originalmente estruturada com metabólitos organizados em linhas e amostras distribuídas em colunas. No entanto, algoritmos de aprendizado de máquina exigem uma estrutura em que cada linha represente uma amostra e cada coluna represente uma variável preditora.

Dessa forma, realiza-se a transposição da tabela, invertendo linhas e colunas.

```python
# Inverte linhas e colunas da tabela
metabolites_ml = metabolites_ml.T

metabolites_ml.head()
```

## Conversão do identificador das amostras em coluna

Após a transposição da tabela, os identificadores das amostras passam a ocupar o índice da estrutura de dados. Entretanto, para permitir a integração com a base clínica, esse índice deve ser convertido em uma coluna explícita, denominada `sample_id`.

```python
# Converte o índice em coluna
metabolites_ml = metabolites_ml.reset_index()

# Renomeia a coluna de índice
metabolites_ml = metabolites_ml.rename(
    columns={'index': 'sample_id'}
)

metabolites_ml.head()
```

## Padronização dos identificadores das amostras

Os identificadores presentes na base metabolômica possuem o sufixo `_CrAdj`, enquanto a base clínica utiliza apenas o identificador principal da amostra. Assim, remove-se esse trecho para garantir compatibilidade entre os conjuntos de dados.

```python
# Remove o sufixo '_CrAdj' dos identificadores
metabolites_ml['sample_id'] = (
    metabolites_ml['sample_id']
    .str.replace('_CrAdj', '', regex=False)
)

metabolites_ml.head()
```

## Integração das bases clínica e metabolômica

Após o pré-processamento das informações clínicas e metabolômicas, realiza-se a integração das bases utilizando o identificador único da amostra (`sample_id`). Esse procedimento permite consolidar, em uma única estrutura, tanto os atributos clínicos quanto os valores metabolômicos.

É utilizada uma junção interna (`inner join`), garantindo que apenas amostras presentes simultaneamente nas duas bases sejam mantidas no conjunto final.

```python
# Integra as bases clínica e metabolômica
dataset_final = pd.merge(
    samples_clean,
    metabolites_ml,
    on="sample_id",
    how="inner"
)

print(dataset_final.shape)
dataset_final.head()
```

```python
# 1. Remover a coluna 'sex'
# Justificativa: todos os 29 machos pertencem exclusivamente
# ao grupo Control. Manter essa variável causaria data leakage
# indireto (o modelo aprenderia "macho = Control" por artefato
# do desenho do estudo, não por padrão metabólico real).
# ----------------------------------------------------------
dataset_final = dataset_final.drop(columns=['sex'])
print("Coluna 'sex' removida.")
```

```python
# 2. Tratamento de valores ausentes em 'age'
# Imputação pela mediana (variável contínua), técnica padrão
# para reduzir o impacto de outliers na imputação.
# 1º imputa age
mediana_age = dataset_final['age'].median()
dataset_final['age'] = dataset_final['age'].fillna(mediana_age)


# 3. Criar coluna auxiliar de idade em anos
# A coluna 'age' original está em meses (conforme metadado
# 'Unit = month' do arquivo s_MTBLS2550). Mantemos 'age' em
# meses para o treinamento (sem perda de precisão) e criamos
# 'age_years' apenas para fins de interpretação e visualização
# nos resultados de XAI.
# 2º cria age_years a partir do age já limpo
dataset_final['age_years'] = (dataset_final['age'] / 12).round(1)
```

```python
# 4. Tratamento da coluna 'neutered'
# Valores 'Not Available' (12 casos) são tratados como uma
# categoria própria em vez de removidos ou imputados, pois
# representam ausência de informação clínica, não um valor
# verdadeiramente ausente no sentido estatístico.
# ----------------------------------------------------------
print("\nDistribuição original de 'neutered':")
print(dataset_final['neutered'].value_counts())

# Codificação ordinal/categórica simples
neutered_dummies = pd.get_dummies(dataset_final['neutered'], prefix='neutered')
dataset_final = pd.concat([dataset_final, neutered_dummies], axis=1)
```

```python
# 5. Codificação da variável alvo (tumor_state)
# Conforme decisão metodológica: classificação binária
# Maligno vs Não-Maligno (Control Group + Benign).
# Justificativa: maior relevância clínica (decisão de
# tratamento agressivo) e maior viabilidade estatística
# dado o tamanho amostral reduzido da classe Benign (n=23).
# ----------------------------------------------------------
dataset_final['target'] = dataset_final['tumor_state'].map({
    'Malignant': 1,
    'Benign': 0,
    'Control Group': 0
})

print("\nDistribuição da variável alvo (target):")
print(dataset_final['target'].value_counts())
```

```python
# 6. Identificação das colunas de metabólitos
# ----------------------------------------------------------
cols_nao_metabolitos = ['sample_id', 'tumor_state', 'age', 'age_years',
                          'neutered', 'target',
                          'neutered_Intact', 'neutered_Neutered', 'neutered_Not Available']
metab_cols = [c for c in dataset_final.columns if c not in cols_nao_metabolitos]
print(f"\nNúmero de metabólitos: {len(metab_cols)}")
```

```python
# 7. Tratamento de valores ausentes nos metabólitos
# Imputação pela mediana de cada metabólito.
# IMPORTANTE: ao fazer k-fold depois, recalcular a mediana
# apenas com dados de treino de cada fold (evitar leakage).
# ----------------------------------------------------------
nulos_antes = dataset_final[metab_cols].isnull().sum().sum()
for col in metab_cols:
    mediana_col = dataset_final[col].median()
    dataset_final[col] = dataset_final[col].fillna(mediana_col)
nulos_depois = dataset_final[metab_cols].isnull().sum().sum()

print(f"\nValores nulos em metabólitos antes: {nulos_antes}")
print(f"Valores nulos em metabólitos depois: {nulos_depois}")
```

```python
## Importação das bibliotecas necessárias para a validação cruzada,
## normalização, construção dos modelos e cálculo das métricas
# ----------------------------------------------------------
import numpy as np

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
## Preparação das variáveis preditoras (X) e da variável alvo (y)

# A partir do dataset final pré-processado, separam-se as variáveis
# preditoras (metabólitos e variáveis clínicas codificadas) da
# variável alvo (target). Colunas de identificação, colunas
# intermediárias (tumor_state, age_years) e a própria variável
# alvo são excluídas do conjunto de preditores.
# ----------------------------------------------------------
cols_excluir = [
    'sample_id', 'tumor_state', 'age_years', 'neutered', 'target'
]

feature_cols = [c for c in dataset_final.columns if c not in cols_excluir]

X = dataset_final[feature_cols].astype(float)
y = dataset_final['target'].astype(int)

print(f"Shape de X: {X.shape}")
print(f"Shape de y: {y.shape}")
print(f"\nFeatures utilizadas ({len(feature_cols)}):")
print(feature_cols)
```

```python
## Configuração da validação cruzada estratificada

# A validação cruzada estratificada (StratifiedKFold) é utilizada
# para preservar a proporção das classes em cada partição,
# garantindo maior estabilidade nas métricas obtidas diante do
# desbalanceamento leve observado entre as classes maligna e
# não maligna.
# ----------------------------------------------------------

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Verificação da estratificação: conferindo se a proporção de
# classes se mantém semelhante em cada fold
for fold, (train_idx, test_idx) in enumerate(skf.split(X, y)):
    prop_treino = y.iloc[train_idx].mean()
    prop_teste = y.iloc[test_idx].mean()
    print(f"Fold {fold+1} — proporção maligno (treino): {prop_treino:.3f} | "
          f"proporção maligno (teste): {prop_teste:.3f}")
```

```python
## Normalização (padronização) das variáveis dentro de cada fold

# As variáveis numéricas apresentam escalas distintas, especialmente
# os metabólitos. Para os modelos de Regressão Logística e SVM,
# sensíveis à escala das variáveis, aplica-se a padronização Z-score
# por meio do StandardScaler.
#
# Para evitar vazamento de informação (data leakage), o scaler é
# ajustado (fit) exclusivamente com os dados de treinamento de cada
# fold, e os mesmos parâmetros são aplicados (transform) ao
# respectivo conjunto de teste, sem novo ajuste.
#
# O Random Forest não exige padronização, pois realiza divisões
# baseadas em limiares de valores, sendo insensível à escala das
# variáveis. Ainda assim, o mesmo conjunto de dados normalizado
# pode ser utilizado sem prejuízo ao seu desempenho.
# ----------------------------------------------------------


# INICIO DE APENAS UM EXEMPLO, A NORMALIZAÇÃO OFICIAL SERÁ APLICADA NAs PRÓXIMAs CÉLULAs
# Exemplo de demonstração da padronização em um único fold,
# para fins de verificação antes do loop completo de treinamento
train_idx, test_idx = next(skf.split(X, y))

X_train_exemplo, X_test_exemplo = X.iloc[train_idx], X.iloc[test_idx]

scaler_exemplo = StandardScaler()
X_train_scaled_exemplo = scaler_exemplo.fit_transform(X_train_exemplo)
X_test_scaled_exemplo = scaler_exemplo.transform(X_test_exemplo)

print("Média das variáveis no treino após padronização (deve ser ~0):")
print(X_train_scaled_exemplo.mean(axis=0)[:5].round(3))

print("\nDesvio-padrão das variáveis no treino após padronização (deve ser ~1):")
print(X_train_scaled_exemplo.std(axis=0)[:5].round(3))

print("\nMédia das variáveis no teste após padronização (não necessariamente 0,")
print("pois usa os parâmetros calculados no treino):")
print(X_test_scaled_exemplo.mean(axis=0)[:5].round(3))
```

# Regressão Logística


```python
## Treinamento e avaliação — Regressão Logística

# Para cada um dos 5 folds da validação cruzada estratificada, os
# dados de treino e teste são separados, a padronização é ajustada
# exclusivamente com os dados de treino e aplicada ao teste, e o
# modelo de Regressão Logística é treinado e avaliado.
# ----------------------------------------------------------
resultados_rl = {'acc': [], 'pre': [], 'rec': [], 'f1': [], 'auc': []}

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y)):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # Padronização ajustada apenas com o treino do fold
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    modelo_rl = LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    )
    modelo_rl.fit(X_train_scaled, y_train)

    y_pred = modelo_rl.predict(X_test_scaled)
    y_prob = modelo_rl.predict_proba(X_test_scaled)[:, 1]

    acc = accuracy_score(y_test, y_pred)
    pre = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc = roc_auc_score(y_test, y_prob)

    resultados_rl['acc'].append(acc)
    resultados_rl['pre'].append(pre)
    resultados_rl['rec'].append(rec)
    resultados_rl['f1'].append(f1)
    resultados_rl['auc'].append(auc)

    print(f"Fold {fold+1} — ACC: {acc:.3f} | PRE: {pre:.3f} | "
          f"REC: {rec:.3f} | F1: {f1:.3f} | AUC: {auc:.3f}")

print(f"\nMédia — ACC: {np.mean(resultados_rl['acc']):.3f} ± {np.std(resultados_rl['acc']):.3f}")
print(f"Média — AUC: {np.mean(resultados_rl['auc']):.3f} ± {np.std(resultados_rl['auc']):.3f}")
```

```python
## Curvas ROC dos três modelos (previsões agregadas dos 5 folds)

# Para construir a curva ROC de cada modelo, as previsões de
# probabilidade obtidas em cada um dos 5 folds da validação
# cruzada são concatenadas, formando um conjunto de previsões
# out-of-fold que cobre a totalidade das 166 amostras da base,
# sem que nenhuma amostra tenha sido utilizada simultaneamente
# para treino e teste.
# ----------------------------------------------------------
from sklearn.metrics import roc_curve

def obter_previsoes_out_of_fold(modelo_classe, modelo_kwargs):
    y_true_completo = []
    y_prob_completo = []

    for fold, (train_idx, test_idx) in enumerate(skf.split(X, y)):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        scaler = StandardScaler()
        X_train_scaled = scaler.fit_transform(X_train)
        X_test_scaled = scaler.transform(X_test)

        modelo = modelo_classe(**modelo_kwargs)
        modelo.fit(X_train_scaled, y_train)
        y_prob = modelo.predict_proba(X_test_scaled)[:, 1]

        y_true_completo.extend(y_test.tolist())
        y_prob_completo.extend(y_prob.tolist())

    return np.array(y_true_completo), np.array(y_prob_completo)

# Obtendo previsões out-of-fold para os três modelos
y_true_rl, y_prob_rl = obter_previsoes_out_of_fold(
    LogisticRegression, {'max_iter': 1000, 'class_weight': 'balanced', 'random_state': 42}
)
y_true_rf, y_prob_rf = obter_previsoes_out_of_fold(
    RandomForestClassifier, {'n_estimators': 100, 'class_weight': 'balanced', 'random_state': 42}
)
y_true_svm, y_prob_svm = obter_previsoes_out_of_fold(
    SVC, {'kernel': 'rbf', 'class_weight': 'balanced', 'probability': True, 'random_state': 42}
)
```

```python
## Plotagem das curvas ROC

fig, ax = plt.subplots(figsize=(7, 7))

for y_true, y_prob, nome, cor in [
    (y_true_rl, y_prob_rl, 'Regressão Logística', '#4C72B0'),
    (y_true_rf, y_prob_rf, 'Random Forest', '#55A868'),
    (y_true_svm, y_prob_svm, 'SVM', '#C44E52'),
]:
    fpr, tpr, _ = roc_curve(y_true, y_prob)
    auc_agregado = roc_auc_score(y_true, y_prob)
    ax.plot(fpr, tpr, label=f'{nome} (AUC = {auc_agregado:.3f})', color=cor, linewidth=2)

ax.plot([0, 1], [0, 1], linestyle='--', color='gray', label='Classificador aleatório (AUC = 0.5)')

ax.set_xlabel('Taxa de falsos positivos')
ax.set_ylabel('Taxa de verdadeiros positivos (Revocação)')
ax.set_title('Curvas ROC dos modelos avaliados')
ax.legend(loc='lower right')
ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig('../figuras/curvas_roc_modelos.png', dpi=150)
plt.show()
```

```python
## Treinamento do modelo final para fins de interpretabilidade (XAI)

# Diferentemente dos modelos avaliados na validação cruzada
# (Seção de Construção dos Modelos), cujo objetivo é estimar
# honestamente a capacidade de generalização por meio de dados
# de teste nunca vistos durante o treinamento, o modelo final
# utilizado nesta seção é treinado com a totalidade das 166
# amostras disponíveis. Essa abordagem é adequada para fins de
# interpretabilidade, uma vez que Feature Importance, SHAP e LIME
# não reportam nenhuma métrica de desempenho preditivo, mas
# descrevem o comportamento interno do modelo já treinado,
# permitindo aproveitar a totalidade da informação disponível
# na base de dados.
#
# O Random Forest foi selecionado para as análises de XAI por
# apresentar a maior estabilidade de desempenho entre as
# partições da validação cruzada (menor desvio-padrão de AUC),
# além de oferecer suporte nativo ao cálculo de Feature Importance.
# ----------------------------------------------------------

# Imputação da mediana em todo o conjunto de dados
# (aqui não há necessidade de recalcular por fold, pois não
# há mais divisão treino/teste nesta etapa)
X_completo = X.copy()
X_completo = X_completo.fillna(X_completo.median())

# Padronização com todo o conjunto de dados
scaler_final = StandardScaler()
X_completo_scaled = scaler_final.fit_transform(X_completo)

# Treinamento do modelo final de Random Forest
modelo_rf_final = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42
)
modelo_rf_final.fit(X_completo_scaled, y)

print("Modelo final treinado com sucesso.")
print(f"Número de amostras utilizadas: {X_completo.shape[0]}")
print(f"Número de variáveis utilizadas: {X_completo.shape[1]}")
```
